```python
import numpy as np

# OR gatetruth table
X = np.array([[0, 0],
             [0, 1],
             [1, 0],
             [1, 1]])
Y = np.array ([[0],
              [1],
              [1],
              [1]])

# Four inputs, each paif of bits. Four expected outputs.
# That's the entire dataset -  no noise, no ambiguity, so if neuron doesn't learn it, the bug is in our math!! 
# not in the data.
```

```python
# Initialize weights and bias. 
np.random.seed(42)  # so we get the same "random" numbers every run — reproducibility matters when comparing notebooks

weights = np.random.randn(2, 1)  # shape (2,1) — one weight per input feature
bias = np.random.randn(1)        # a single scalar bias

print("weights:\n", weights)
print("bias:", bias)
```

```python
# The forward pass (weighted sum + Activation function)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def forward(X, weights, bias):
    z = np.dot(X, weights) + bias   # weighted sum: (4,2) @ (2,1) -> (4,1), then add bias
    a = sigmoid(z)                   # squash into (0,1) range
    return a

predictions = forward(X, weights, bias)
print(predictions)
```

```python
# The loss function (Mean Squared Error)
def mse_loss(Y_true, y_pred):
    return np.mean((Y_true - y_pred) ** 2)

loss = mse_loss(Y, predictions)
print("Loss:", loss)
```

```python
# Backpropagation (Computing gradients manually)
def backward(X, Y, a):
    n = Y.shape[0]
    
    dL_da = (2 / n) * (a - Y)          # Step 1: loss w.r.t. prediction
    da_dz = a * (1 - a)                 # Step 2: sigmoid derivative
    dL_dz = dL_da * da_dz               # Step 3: chain rule combines them
    
    dL_dw = np.dot(X.T, dL_dz)          # Step 4a: gradient w.r.t. weights
    dL_db = np.sum(dL_dz)                # Step 4b: gradient w.r.t. bias
    
    return dL_dw, dL_db

dW, dB = backward(X, Y, predictions)
print("Gradient for weights:\n", dW)
print("Gradient for bias:", dB)
```

Gradient descent update + the training loop

Why: We now have gradients (dW, dB) telling us which direction makes the loss worse. Gradient descent just means: move the opposite direction, by a small amount. That "small amount" is controlled by a number called the learning rate — too big and the neuron overshoots and never settles; too small and it learns painfully slowly.

The update rule is simple:

weights = weights - learning_rate * dW
bias = bias - learning_rate * dB

```python
learning_rate = 0.1
epochs = 10000

for epoch in range(epochs):
    # forward pass
    predictions = forward(X, weights, bias)
    
    # compute loss
    loss = mse_loss(Y, predictions)
    
    # backward pass
    dW, dB = backward(X, Y, predictions)
    
    # update weights and bias
    weights = weights - learning_rate * dW
    bias = bias - learning_rate * dB
    
    # print progress occasionally
    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
# Test the trained neuron
final_predictions = forward(X, weights, bias)

print("Final predictions:")
print(final_predictions)

print("\nRounded (0 or 1):")
print(np.round(final_predictions))

print("\nActual Y:")
print(Y)
```
